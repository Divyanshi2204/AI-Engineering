```python
import requests
from bs4 import BeautifulSoup

url = "https://www.passiton.com/inspirational-quotes"
response = requests.get(url)
soup = BeautifulSoup(response.text, 'html.parser')

quotes = []

# Updated class name to match the website's new structure
quote_boxes = soup.find_all('div', class_='text-center mb-8')

for box in quote_boxes:
    if not box.img:
        continue
        
    p_tags = box.find_all('p')
    # Extract lines and author from paragraph tags, falling back to image alt tags if needed
    quote_text = box.img.get('alt', '').split(" #")
    lines = p_tags[0].text.strip('“”‘’"\'') if len(p_tags) > 0 else quote_text[0]
    author = p_tags[1].text.strip() if len(p_tags) > 1 else (quote_text[1] if len(quote_text) > 1 else 'Unknown')
    
    quote = {
        'theme': box.h5.text.strip() if box.h5 else 'General',
        'image_url': box.img['src'],
        'lines': lines,
        'author': author
    }
    quotes.append(quote)

print(f"Successfully scraped {len(quotes)} quotes.")
```

```python
from IPython.display import display, HTML

# Generate HTML for each quote card
html_content = "<div style='display: flex; flex-wrap: wrap; gap: 20px;'>"
for q in quotes[:8]:  # Showing first 8 quotes
    img_url = q['image_url']
    if img_url.startswith('/'):
        img_url = "https://www.passiton.com" + img_url
        
    html_content += f"""
    <div style="border: 1px solid #ccc; border-radius: 8px; padding: 15px; width: 220px; box-shadow: 2px 2px 5px rgba(0,0,0,0.1); background-color: #fff; text-align: center;">
        <h4 style="color: #007bff; margin-top: 0; text-transform: uppercase; font-size: 14px;">{q['theme']}</h4>
        <img src="{img_url}" alt="{q['theme']}" style="width: 100%; border-radius: 4px; margin-bottom: 10px;" />
        <p style="font-style: italic; font-size: 14px; color: #333;">"{q['lines']}"</p>
        <p style="text-align: right; font-weight: bold; color: #555; font-size: 12px; margin-top: 5px;">— {q['author']}</p>
    </div>
    """
html_content += "</div>"

display(HTML(html_content))
```
